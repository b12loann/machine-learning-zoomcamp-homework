```python
del df_train['msrp']
del df_val['msrp']
del df_test['msrp']
```

# Find most important features

```python
global_msrp = df_full_train.msrp.mean()
global_msrp
```

```python
from sklearn.metrics import mutual_info_score
```

```python
def mutual_info_msrp_score(series):
    return mutual_info_score(series, df_full_train.msrp)
```

```python
df_full_train.dtypes
```

```python
df_full_train.columns
```

```python
categorical = ['make', 'model', 'engine_fuel_type', 'transmission_type', 'driven_wheels', 'market_category', 'vehicle_size', 'vehicle_style']
```

```python
numerical = ['year', 'engine_hp', 'engine_cylinders', 'number_of_doors', 'highway_mpg', 'city_mpg', 'popularity']
```

```python
from IPython.display import display
```

```python
for c in categorical:
    df_group = df_full_train.groupby(c).msrp.agg(['mean', 'count'])
    df_group['diff'] = df_group['mean'] - global_msrp
    df_group['risk'] = df_group['mean']/ global_msrp
    display(df_group)
    print()
    print()
```

```python
df_full_train[categorical].dtypes
```

```python
df_full_train[categorical]
```

```python
mi =df_full_train[categorical].apply(mutual_info_msrp_score)
mi.sort_values(ascending=False)
```

```python
categorical_important = ['make', 'model', 'engine_fuel_type', 'transmission_type', 'market_category', 'vehicle_style']
```

```python
numerical = ['year', 'engine_hp', 'engine_cylinders', 'number_of_doors', 'highway_mpg', 'city_mpg', 'popularity']
```

# One-hot encoding

```python
from sklearn.feature_extraction import DictVectorizer
```

## Train dataset

```python
train_dicts = df_train[categorical_important + numerical_important].to_dict(orient="records")
```

```python
dv = DictVectorizer(sparse=False)
```

```python
dv.fit(train_dicts)
```

```python
# do directily fit & transform
X_train = dv.fit_transform(train_dicts)
X_train
```

```python
X_train.shape
```

## Val dataset

```python
val_dicts = df_val[categorical_important + numerical_important].to_dict(orient="records")
```

```python
dv.fit(val_dicts)
```

```python
X_val = dv.transform(val_dicts)
X_val
```

# Linear regression

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_val)
```

```python
y_pred >= 0.5
```

```python
msrp_decision = y_pred >= 0.5
msrp_decision
```
